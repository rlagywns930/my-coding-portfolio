- df2의 업체명들의 중복한 것들은 / 를 해서 하나의 데이터로 모아준다. 그리고 중복된 것을 없애주고 문자열로 바꿔준다.
- 각 칼럼마다 적절한 방법으로 적용해 준다. 중복이 되는것은 첫번째만 보여주고 수를 말하는 컬럼은 더해주고 범위를 나타내면 첫번쨰만 보여준다.

````python
route_info = (
    df2.groupby(['사용년월','노선번호'],as_index=False)
       .agg(
           업체명=(
               '업체명', lambda x: ' / '.join(x.dropna().astype(str).drop_duplicates())),
           유형=('유형', 'first'),
           기점=('기점', 'first'),
           종점=('종점', 'first'),

           인가대수=('인가대수', 'sum'),
           운행대수=('운행대수', 'sum'),
           예비대수=('예비대수', 'sum'),
           총운행횟수=('총운행횟수', 'sum'),

           인가거리=('인가거리', 'first'),
           운행시간=('운행시간', 'first'),

           최소=('최소', 'min'),
           최대=('최대', 'max'),

           첫차시간=('첫차시간', 'first'),
           막차시간=('막차시간', 'first')
       ))
````

- 첫차 시간과 막차시간을 시간형식으로 바꿔준다. (시,분)

````python
def convert_bus_time(series):
    
    time_text = pd.to_numeric(series, errors='coerce').astype('Int64').astype('string').str.zfill(4)
    
    return pd.to_datetime(time_text, format='%H%M', errors='coerce')
````

````python
analysis_df['첫차시간_변환'] = convert_bus_time(analysis_df['첫차시간'])

analysis_df['막차시간_변환'] = convert_bus_time(analysis_df['막차시간'])
````

````python
analysis_df['첫차시간_표시'] = analysis_df['첫차시간_변환'].dt.strftime('%H:%M')

analysis_df['막차시간_표시'] = analysis_df['막차시간_변환'].dt.strftime('%H:%M')
````

````python
# 총 승차 인원을 구할 것입니다.

ride_cols = [col for col in analysis_df .columns if '승차총승객수' in col]

analysis_df [ride_cols] = analysis_df [ride_cols].apply(pd.to_numeric, errors='coerce').fillna(0)

analysis_df ['총승차인원'] = analysis_df [ride_cols].sum(axis=1)
````

````python
# 2025년 7월~2026년 6월 공급 지표 평균을 계산합니다.

supply_h1_df=positive_route_info.groupby('노선번호',as_index=False).agg(
    업체명=('업체명','first'),
    유형=('유형','first'),
    기점=('기점','first'),
    종점=('종점','first'),
    인가대수=('인가대수','mean'),
    운행대수=('운행대수','mean'),
    예비대수=('예비대수','mean'),
    총운행횟수=('총운행횟수','mean'),
    최소배차=('최소','mean'),
    최대배차=('최대','mean'),
    인가거리=('인가거리','mean'),
    운행시간=('운행시간','mean')
)
supply_h1_df['평균운행횟수']=supply_h1_df['총운행횟수']/supply_h1_df['운행대수']
````

````python
## 현재 데이터들은 정류장으로 구분이 되어있는데 두개의 데이터 베이스를 각각 노선으로 묶어줍니다

time_cols=[col for col in top_low_ride_df.columns if col.endswith('시승차인원')]

def make_route_summary(data):
    route_summary = data.groupby(['노선번호','구분'],as_index=False).agg(
        노선명 = ('노선명','first'),
        유형 = ('유형','first'),
        노선전체승차인원합계 = ('정류장승차인원합계','sum'),
        정류장수 = ('표준버스정류장ID','nunique'),
        운행대수 = ('운행대수','first'),
        총운행횟수 = ('총운행횟수','first'),
        평균운행횟수 = ('평균운행횟수','first'),
        인가거리 = ('인가거리','first'),
        운행시간 = ('운행시간','first'),
        최소배차 = ('최소배차','first'),
        최대배차 = ('최대배차','first')
    )

    time_summary = data.groupby(['노선번호','구분'],as_index=False)[time_cols].sum()
    route_summary = route_summary.merge(time_summary,on=['노선번호','구분'])

    route_summary['월평균승차인원'] = route_summary['노선전체승차인원합계'] / 12
    route_summary['차량1대당월평균승차인원'] = route_summary['월평균승차인원'] / route_summary['운행대수']

    return route_summary
````

````python
total_route_summary_df = make_route_summary(top_low_ride_df)

limit_route_summary_df = make_route_summary(top_low_limit_ride_df)
````

````python
# 다음 단계: 시간대 이용 특성 추가

time_cols = ['01~03시승차인원', '04~06시승차인원', '07~09시승차인원', '10~12시승차인원', '13~15시승차인원', '16~18시승차인원', '19~21시승차인원', '22~24시승차인원']

def add_time_features(data):
    
    result = data.copy()

    total = (result['노선전체승차인원합계'].replace(0, np.nan))

    result['출근시간비율'] = (result['07~09시승차인원'] / total)

    result['퇴근시간비율'] = (result[['16~18시승차인원','19~21시승차인원']].sum(axis=1) / total)

    result['심야시간비율'] = (result[['22~24시승차인원', '01~03시승차인원']].sum(axis=1) / total)

    ############ 시간대의 문자열만 가져옵니다.

    result['주이용시간대'] = (result[time_cols].idxmax(axis=1).str.replace('승차인원', '', regex=False))

    result['최대시간대집중도'] = (result[time_cols].max(axis=1) / total)

    return result
````

````python
# 두개의 데이터의 적용

total_route_summary_df = add_time_features(total_route_summary_df)

limit_route_summary_df = add_time_features(limit_route_summary_df)
````

# 유형별
- 간선·광역·맞춤·지선 모두 상위 노선의 월평균 운행대수와 총운행횟수가 더 많았다.
- 차량 1대당 평균운행횟수는 유형별로 방향이 달랐다. 광역은 상위가 더 높지만, 간선·맞춤·지선은 하위가 더 높았다.
- 상위 노선의 배차간격은 비교 가능한 유형에서 대체로 더 짧았다. 다만 맞춤버스의 최대배차 차이는 크지 않았다.
- 순환과 심야는 비교할 하위 노선이 없어 상·하위 차이를 판단할 수 없다.
- 간선·광역·맞춤·지선 모두 상위 노선의 정류장수, 인가거리, 운행시간이 더 컸다.


````python
# 노선 규모를 고려하기 위해 거리당 월평균 승차인원을 계산합니다.

route_time_analysis_df['거리당월평균승차인원']=route_time_analysis_df['월평균승차인원']/route_time_analysis_df['인가거리'].replace(0,np.nan)
````

````python
#유형별 상·하위 중앙값을 비교합니다.
scale_cols=['정류장당월평균승차인원','거리당월평균승차인원']

scale_comparison_df = route_time_analysis_df[route_time_analysis_df['유형내구분']!='중간'].groupby\
(['유형','유형내구분'])[scale_cols].median().round(1)

scale_comparison_df


##### 정류장당월평균승차인원: 월평균승차인원 ÷ 정류장수
##### 거리당월평균승차인원: 월평균승차인원 ÷ 인가거리
````

- 상위 노선의 정류장당월평균승차인원이 많은 것은 단순히 노선이 길거나 정류장이 많기 때문만은 아니다. 정류장수와 노선거리를 고려한 후에도 같은 유형의 상위 노선에서 승차인원이 더 많았다.
- 상위 노선은 하위 노선보다 정류장수가 많고 운행거리도 길어 더 많은 이용 기회를 가진다. 동시에 정류장당·거리당 승차인원도 더 높아, 이용량 차이는 노선 규모뿐 아니라 단위 구간의 이용 수요 차이와도 연결되어 있다.

````python
# 1. 노선·정류장별 승차인원 집계

route_stop_df = analysis_df.groupby(['노선번호', '표준버스정류장ID'], as_index=False).agg(역명=('역명', 'first'), 정류장전체승차인원=('총승차인원', 'sum'))
route_stop_df
````

````python
# 2. 노선 전체 승차인원 추가
route_stop_df['노선전체승차인원'] = route_stop_df.groupby('노선번호')['정류장전체승차인원'].transform('sum')

route_stop_df['정류장승차비율'] = route_stop_df['정류장전체승차인원']/route_stop_df['노선전체승차인원']
route_stop_df
````

````python
# 3. 노선별 정류장 집중도 계산
# 최대정류장집중도: 이용량이 가장 많은 정류장 1개의 비중
# 상위5정류장집중도: 이용량이 많은 정류장 5개의 비중 합계


route_stop_concentration_df = route_stop_df.groupby('노선번호').agg(최대정류장집중도=('정류장승차비율', 'max'), 상위5정류장집중도=('정류장승차비율', lambda x: x.nlargest(5).sum())).reset_index()
route_stop_concentration_df
````

````python
# 4. 기존 노선 분석 데이터와 결합
route_concentration_df = route_time_analysis_df.merge(route_stop_concentration_df, on='노선번호', how='left')
route_concentration_df
````

````python
# 5. 같은 유형의 상·하위 비교
stop_comparison_df = route_concentration_df[route_concentration_df['유형내구분'] != '중간'].groupby(['유형', '유형내구분'])[['최대정류장집중도', '상위5정류장집중도']].median().mul(100).round(1)
stop_comparison_df
````

하차 수요가 집중된 노선에 버스 공급이 부족한가?

1. 왜 유형별 평균을 구했나?
- 간선과 지선은 원래 운행대수·운행횟수·배차가 다를 수 있습니다. 그래서 간선은 전체 간선 평균과, 지선은 전체 지선 평균과 비교했습니다.
- 비교한 공급 변수는 세 개입니다.
- 운행대수: 투입된 차량 수
- 총운행횟수: 하루 전체 운행횟수
- 추정대표배차: 버스가 오는 추정 간격


````python
# 운행대수 0 노선을 제외하여 만든 12개월 노선별 평균 공급정보를 사용합니다.

all_supply_df=supply_h1_df.copy()

all_supply_df['차량당일운행횟수']=all_supply_df['총운행횟수']/all_supply_df['운행대수']

all_supply_df['추정대표배차']=(all_supply_df['최소배차']+all_supply_df['최대배차'])/2
````

````python
## 유형별 공급 평균

type_supply_avg_df = all_supply_df.groupby('유형',as_index=False).agg(
    유형노선수=('노선번호','nunique'),
    유형평균운행대수=('운행대수','mean'),
    유형평균총운행횟수=('총운행횟수','mean'),
    유형평균차량당일운행횟수=('차량당일운행횟수','mean'),
    유형평균추정대표배차=('추정대표배차','mean')
    )
````

````python
## 유형별 평균 붙이기

supply_compare_df = daily_supply_analysis_df.merge(type_supply_avg_df,on='유형',how='left')
````

- . 세 개의 True·False는 무엇인가?
1. 운행대수평균미만
- 해당 노선의 운행대수가 같은 유형 평균보다 적으면 True입니다.
2. 총운행횟수평균미만
- 해당 노선의 일일 총운행횟수가 같은 유형 평균보다 적으면 True입니다.
3. 대표배차평균초과
- 해당 노선의 배차간격이 같은 유형 평균보다 길면 True입니다.
- 즉, 모두 True일수록 공급 조건이 불리하다는 뜻입니다.

````python
## 공급이 같은 유형의 평균보다 불리한지 확인

supply_compare_df['운행대수평균미만'] = supply_compare_df['운행대수'] < supply_compare_df['유형평균운행대수']
supply_compare_df['총운행횟수평균미만'] = supply_compare_df['총운행횟수'] < supply_compare_df['유형평균총운행횟수']
supply_compare_df['대표배차평균초과'] = supply_compare_df['추정대표배차'] > supply_compare_df['유형평균추정대표배차']
````

````python
## 세 조건 중 몇 개가 불리한지 계산합니다. True를 1, False를 0으로 바꿔 더하는 것

supply_compare_df['공급불리항목수'] = supply_compare_df['운행대수평균미만'].astype(int) + supply_compare_df['총운행횟수평균미만'].astype(int) + supply_compare_df['대표배차평균초과'].astype(int)
````

````python
## 세 조건 중 두 개 이상 불리한 노선을 후보로 뽑는다.

supply_candidate_df = supply_compare_df[supply_compare_df['공급불리항목수'] >= 2].copy()
supply_candidate_df = supply_candidate_df.sort_values(['공급불리항목수','추정회당집중시간대하차인원'],ascending=[False,False])
````

- 왜 공급불리항목수를 만들었나?
1. True가 몇 개인지 세기 위해 만들었습니다.
- 예를 들어 5515번은:
- 운행대수평균미만: True

### 종합 결론

서울시 상위 승차 정류장의 수요는 주로 저녁 시간대에 집중되며, 여러 노선으로 이용이 분산되는 가운데 일부 노선이 균등 기준보다 약 `1.69~3.78배` 높은 승차 비중을 보였다.

다만 `최대승차시간대`는 정류장 전체의 시간대이고 `노선승차비율`은 전체 기준이므로, 특정 노선의 비중이 해당 시간대에 실제로 증가했는지는 별도의 시간대별 노선승차비율 분석이 필요하다.

 전체 노선승차비율
vs
정류장 최대 승차시간대의 노선승차비율

- 집중시간대에도 같은 노선이 1위인지
- 특정 시간대에 그 노선의 비중이 더 커지는지
- 저녁 승차 집중이 정류장 전체 현상인지, 특정 노선이 주도하는지

````python
boarding_peak_time_df = top20_boarding_stop_df[['표준버스정류장ID','최대승차시간대']].copy()
top_boarding_peak_df = top_boarding_route_df.merge(boarding_peak_time_df, on= '표준버스정류장ID', how= 'left')
top_boarding_peak_df[['표준버스정류장ID','역명','노선번호','최대승차시간대']].head()
````

- boarding_peak_time_df: 정류장별 최대 승차시간대만 가져옵니다.
- top_boarding_route_df: 한 행이 정류장 × 노선인 데이터입니다.
- merge: 각 노선 행에 해당 정류장의 최대 승차시간대를 붙입니다.
- top_boarding_peak_df: 앞으로 집중시간대의 노선별 승차비율을 계산할 데이터입니다.

````python
# 각 정류장·노선 행에서 최대 승차시간대에 해당하는 승차인원만 가져오는 단계

top_boarding_peak_df['집중시간대승차인원'] = 0

for col in time_group_cols:
    time_name = col.replace('승차인원','')
    mask = top_boarding_peak_df['최대승차시간대'] == time_name
    top_boarding_peak_df.loc[mask,'집중시간대승차인원'] = top_boarding_peak_df.loc[mask,col]

top_boarding_peak_df[['표준버스정류장ID','역명','노선번호','최대승차시간대','집중시간대승차인원']].head()
````

- time_name: 16~18시승차인원에서 승차인원을 제거해 16~18시로 만듭니다.
- mask: 최대 승차시간대가 16~18시인 행을 찾습니다.
- .loc: 찾은 행에 해당 시간대의 실제 승차인원을 넣습니다.

````python
# 집중시간대에 해당 정류장의 전체 노선 승차인원을 구하고, 노선별 비율을 계산하는 단계
boarding_peak_total_df = top_boarding_peak_df.groupby('표준버스정류장ID',as_index=False)['집중시간대승차인원'].sum().rename(columns={'집중시간대승차인원':'집중시간대전체승차인원'})
top_boarding_peak_df = top_boarding_peak_df.merge(boarding_peak_total_df, on='표준버스정류장ID', how='left')
top_boarding_peak_df['집중시간대노선승차비율'] = top_boarding_peak_df['집중시간대승차인원'] / top_boarding_peak_df['집중시간대전체승차인원'] * 100
top_boarding_peak_df
````

- 특정 노선의 집중시간대 승차인원 ÷ 같은 정류장 모든 노선의 집중시간대 승차인원 * 100

````python
peak_dominant_boarding_df = top_boarding_peak_df.sort_values('집중시간대노선승차비율',ascending=False).drop_duplicates('표준버스정류장ID').copy()
peak_dominant_boarding_df[['표준버스정류장ID','역명','최대승차시간대','노선번호','집중시간대노선승차비율']].sort_values('집중시간대노선승차비율',ascending=False).round(2)
````

## 8. 분석의 한계와 개선 방향

 **08~09시에는 하차 수요와 회당 하차인원이, 18~19시에는 승차 수요와 회당 승차인원이 함께 높게 나타났으므로**, 해당 시간대의 실제 혼잡도와 운행 안정성을 우선적으로 확인할 필요가 있다.

추가 조사에서 혼잡, 미탑승 또는 지연이 확인된다면 해당 시간대의 운행횟수 조정이나 배차 개선을 검토할 수 있다.

### 정류장 순서 기반 추정 재차인원 분석

정류장 순서에 따라 각 정류장의 회당 승차인원과 회당 하차인원의 차이를 누적하여 노선 안에서 수요가 쌓이고 감소하는 흐름을 확인한다.

신동아아파트 종점은 승·하차 자료에서 `00001`과 `00033` 두 가상 정류장으로 나뉜다. `00001`은 출발 측 순서 0, `00033`은 도착 측 마지막 순서 32로 배치한다. 두 가상 정류장에는 실제 운행횟수가 기록되지 않아 각각 인접한 첫 번째·마지막 실제 정류장의 운행횟수를 적용한다.

````python
# 정류장 순서와 실제 승·하차 숫자를 이용해 시간대별 추정 재차인원을 계산합니다.
route_5515_load_df = route_5515_hourly_df.copy()

# 신동아아파트 종점의 출발 측 가상 정류장과 도착 측 가상 정류장을 노선 순서에 배치합니다.
start_virtual_id = 990070751  # 신동아아파트(가상)(00001)
end_virtual_id = 990070766    # 신동아아파트(가상)(00033)
first_real_id = 120000225     # 금호타운아파트(00002)
last_real_id = 120000220      # 금호타운아파트(00032)

route_5515_load_df.loc[route_5515_load_df['표준버스정류장ID'] == start_virtual_id, '정류장_순서'] = 0
route_5515_load_df.loc[route_5515_load_df['표준버스정류장ID'] == end_virtual_id, '정류장_순서'] = 32

# 가상 정류장은 실제 운행횟수가 0이므로 인접한 실제 정류장의 시간대별 운행횟수를 적용합니다.
first_stop_runs = route_5515_load_df[route_5515_load_df['표준버스정류장ID'] == first_real_id].set_index('시간대')['실제운행횟수']
last_stop_runs = route_5515_load_df[route_5515_load_df['표준버스정류장ID'] == last_real_id].set_index('시간대')['실제운행횟수']

start_mask = route_5515_load_df['표준버스정류장ID'] == start_virtual_id
end_mask = route_5515_load_df['표준버스정류장ID'] == end_virtual_id
route_5515_load_df.loc[start_mask, '실제운행횟수'] = route_5515_load_df.loc[start_mask, '시간대'].map(first_stop_runs)
route_5515_load_df.loc[end_mask, '실제운행횟수'] = route_5515_load_df.loc[end_mask, '시간대'].map(last_stop_runs)

route_5515_load_df = route_5515_load_df.sort_values(['시간대','정류장_순서','표준버스정류장ID']).copy()
operation_denominator = route_5515_load_df['실제운행횟수'].replace(0, np.nan)
route_5515_load_df['회당승차인원'] = route_5515_load_df['승차인원'] / operation_denominator
route_5515_load_df['회당하차인원'] = route_5515_load_df['하차인원'] / operation_denominator
route_5515_load_df['회당순증가'] = route_5515_load_df['회당승차인원'] - route_5515_load_df['회당하차인원']

# 집계자료의 시간대 경계 차이로 생기는 잔여 음수는 실제 승객 수가 될 수 없으므로 0을 하한으로 누적합니다.
def accumulate_nonnegative(series):
    current_load = 0.0
    values = []
    for net_change in series.fillna(0.0):
        current_load = max(0.0, current_load + float(net_change))
        values.append(current_load)
    return pd.Series(values, index=series.index)

route_5515_load_df['보정전누적값'] = route_5515_load_df.groupby('시간대')['회당순증가'].cumsum()
route_5515_load_df['추정재차인원'] = route_5515_load_df.groupby('시간대')['회당순증가'].transform(accumulate_nonnegative)
````

````python
## 오전 07~09시와 비교 시간대 13시의 최대 추정재차인원

analysis_hours = ['07시','08시','09시','13시']
route_5515_load_df[route_5515_load_df['시간대'].isin(analysis_hours)][
    ['시간대','정류장_순서','역명','회당승차인원','회당하차인원','추정재차인원']
].sort_values('추정재차인원',ascending=False).head(20).round(2)
````

#### 계산 방법

- 회당 승차인원 = 정류장별 승차인원 ÷ 실제 운행횟수
- 회당 하차인원 = 정류장별 하차인원 ÷ 실제 운행횟수
- 회당 순증가 = 회당 승차인원 － 회당 하차인원
- 추정재차인원 = 정류장 순서에 따른 회당 순증가의 누적값

신동아아파트 출발 측 가상 정류장의 승차 수요를 순서 0에 포함했으며, 집계자료의 차이로 누적값이 0보다 작아지는 경우에는 물리적으로 가능한 최솟값인 0을 적용했다.

#### 해석 범위

이 계산은 정류장 순서와 승·하차 숫자 데이터를 결합한 **시간대별 평균 재차인원 추정값**이다. 노선 앞쪽에서 수요가 누적되고 주요 하차 정류장에서 감소하는 흐름을 비교하는 데 사용한다.

다만 정류장별로 집계된 여러 날짜와 여러 차량의 평균을 연결한 값이므로 특정 차량의 실제 탑승 인원이나 차량 정원 대비 혼잡률과 동일하지 않다. 음수 방지 보정이 적용된 구간은 실제 혼잡도 결론에 사용하지 않는다.

````python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

boarding_cols = [
    col for col in route_5515_time_df.columns
    if '승차총승객수' in col
]

alighting_cols = [
    col for col in route_5515_time_df.columns
    if '하차총승객수' in col
]

boarding_cols = sorted(
    boarding_cols,
    key=lambda x: int(x.split('시')[0])
)
alighting_cols = sorted(
    alighting_cols,
    key=lambda x: int(x.split('시')[0])
)

boarding = route_5515_time_df[boarding_cols].sum()
alighting = route_5515_time_df[alighting_cols].sum()

hours = [
    f'{i}시'
    for i in range(24)
]

plt.figure(figsize=(15,7))
plt.plot(
    range(24),
    boarding.values,
    marker='o',
    linewidth=2.5,
    markersize=6,
    color='#D9534F',
    label='승차'
)
plt.plot(
    range(24),
    alighting.values,
    marker='s',
    linewidth=2.5,
    markersize=6,
    color='#4A90E2',
    label='하차'
)
b_idx = np.argmax(
    boarding.values
)
plt.scatter(
    b_idx,
    boarding.values[b_idx],
    s=160,
    color='#D9534F',
    zorder=5
)
plt.annotate(
    f'승차 최대\n{hours[b_idx]}\n({boarding.values[b_idx]:,}명)',
    (b_idx, boarding.values[b_idx]),
    xytext=(40,20),
    textcoords='offset points',
    ha='center',
    color='#D9534F',
    fontweight='bold'
)
a_idx = np.argmax(
    alighting.values
)
plt.scatter(
    a_idx,
    alighting.values[a_idx],
    s=160,
    color='#4A90E2',
    zorder=5
)
plt.annotate(
    f'하차 최대\n{hours[a_idx]}\n({alighting.values[a_idx]:,}명)',
    (a_idx, alighting.values[a_idx]),
    xytext=(-50,20),
    textcoords='offset points',
    ha='center',
    color='#4A90E2',
    fontweight='bold'
)


plt.title(
    '5515 노선 시간대별 승·하차 인원 (2025.07~2026.06)',
    fontsize=16,
    fontweight='bold',
    pad=25
)
plt.xlabel(
    '시간대',
    fontsize=12
)
plt.ylabel(
    '승객 수',
    fontsize=12
)

max_y = max(
    boarding.max(),
    alighting.max()
)
min_y = min(
    boarding.min(),
    alighting.min()
)


plt.ylim(
    min_y - (max_y-min_y)*0.05,
    max_y*1.18
)
plt.xlim(
    -0.5,
    23.5
)
plt.xticks(
    range(24),
    hours
)
plt.grid(
    axis='y',
    linestyle='--',
    alpha=0.4
)
plt.legend()
plt.tight_layout()
plt.show()
````

````python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

def get_station_value(df, station_name):
    data = df[df['역명'].str.contains(station_name, na=False)]
    
    if len(data) == 0:
        return 0, 0
    
    boarding = data['전체07_09승차인원'].sum()
    alighting = data['전체07_09하차인원'].sum()
    
    return boarding, alighting

g_5515_up, g_5515_down = get_station_value(
    route_5515_flow_df,
    '금호타운아파트'
)

s_5515_up, s_5515_down = get_station_value(
    route_5515_flow_df,
    '서울대입구역2번출구'
)

value_5515 = g_5515_up + s_5515_down

g_8552_up, g_8552_down = get_station_value(
    route_8552_stop_df,
    '우림시장'
)

s_8552_up, s_8552_down = get_station_value(
    route_8552_stop_df,
    '신림사거리'
)

value_8552 = g_8552_up + s_8552_down

compare_df = pd.DataFrame({
    '노선': [
        '8552\n우림시장→신림역',
        '5515\n금호타운→서울대입구역2번출구'
    ],
    '출근 집중 규모': [
        value_8552,
        value_5515
    ]
})

fig, ax = plt.subplots(figsize=(10, 6))

colors = ['#4A90E2', '#D9534F']

bars = ax.bar(
    compare_df['노선'],
    compare_df['출근 집중 규모'],
    color=colors,
    width=0.5,
    edgecolor='white'
)

ax.set_title(
    '출근시간(07~09시) 주요 통근 구간 집중 규모 비교',
    fontsize=16,
    fontweight='bold',
    pad=20
)

ax.set_ylabel('승차 + 하차 인원', fontsize=13)

max_val = max(compare_df['출근 집중 규모'])
ax.set_ylim(0, max_val * 1.15)

for bar, value in zip(bars, compare_df['출근 집중 규모']):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value + (max_val * 0.02),
        f'{int(value):,}명',
        ha='center',
        va='bottom',
        fontsize=12,
        fontweight='bold'
    )

ax.grid(axis='y', linestyle='--', alpha=0.3)

for spine in ['top', 'right', 'left']:
    ax.spines[spine].set_visible(False)

ax.tick_params(axis='both', labelsize=11)

plt.tight_layout()
plt.show()
````

````python
def concentration(df):
    boarding = df.groupby('역명')['전체07_09승차인원'].sum()
    max_board_station = boarding.idxmax()
    board_ratio = boarding.max() / boarding.sum() * 100

    alighting = df.groupby('역명')['전체07_09하차인원'].sum()
    max_alight_station = alighting.idxmax()
    alight_ratio = alighting.max() / alighting.sum() * 100

    return [
        max_board_station,
        round(board_ratio, 1),
        max_alight_station,
        round(alight_ratio, 1)
    ]

result = pd.DataFrame(
    [
        concentration(route_8552_stop_df),
        concentration(route_5515_flow_df)
    ],
    columns=[
        '최대 승차 정류장',
        '승차 집중도(%)',
        '최대 하차 정류장',
        '하차 집중도(%)'
    ],
    index=['8552', '5515']
)

result
````
